```python
# Import libraries
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

#Handle missing value




```python
# Import data set
price_df = pd.read_csv('/content/10_Paris House Price (1).csv')
price_df.head()
```

```python
price_df.info()
```

```python
df = price_df.dropna(subset=['numberOfRooms'])
df.head()
```

delete because the number of missing value is just one

```python
df.info()
```

```python
#chuyển đổi loại dữ liệu cho price
df['price'] = df['price'].str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(float)
df.info()
df.head()
```

```python
df.boxplot(column='price')
```

```python
df.boxplot(column='squareMeters')
#khong có outlier
```

```python
# Check duplicated rows
print(price_df.duplicated().sum())
# No duplicate rows in both datasets
```

```python
print(f'Shape of DataFrame before dropping duplicates: {df.shape}')
```

```python
df.drop_duplicates(inplace=True)
print(f'Shape of DataFrame after dropping duplicates: {df.shape}')
```

```python
df['price_per_squareMeter'] = df['price'] / df['squareMeters']
df.head()
```

```python
current_year = pd.Timestamp.now().year
df['property_age'] = 2026 - price_df['made']
df.head()
```

```python
conditions = [
    df['property_age'] <= 5,
    (df['property_age'] > 5) & (df['property_age'] <= 15),
    df['property_age'] > 15
]
choices = ['new', 'current', 'old']
df['age_category'] = np.select(conditions, choices, default='unknown')
df.head()
```

```python
conditions_size = [
    df['squareMeters'] <= 25000,
    df['squareMeters'] <= 50000,
    df['squareMeters'] <= 75000
]
choices_size = ['small', 'medium', 'large']
df['size_category'] = np.select(conditions_size, choices_size, default='extra large')
df.head()
```

```python
description = df['price'].describe()
print(description.round(3))
```

```python
df.to_csv('price_cleaned(1).csv', index=False)
print('price_cleaned_df saved as price_cleaned.csv')
```

#Insight 1: The Dominance of Square Meters

```python
numeric_df = df.select_dtypes(include=[np.number])
correlation_matrix = numeric_df.corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```

squaremeter và price có hệ số correlation là 1. tương quan positive mạnh mẽ, nghĩa là giá nhà ở hoàn toàn được quyết định bởi yếu tối diện tích

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='squareMeters', y='price', data=df, alpha=0.7)
plt.title('Price vs. Square Meters')
plt.xlabel('Square Meters')
plt.ylabel('Price')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

#Insight 2:The paradox -  Luxury combos don't necessarily drive up housing prices.

```python
df['Luxury_Score'] = df['hasPool'] + df['hasYard'] + df['hasStormProtector'] + df['isNewBuilt']
df.head()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Luxury_Score', y='price', data=df)
plt.title('Price Distribution by Luxury Score')
plt.xlabel('Luxury Score')
plt.ylabel('Price')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

Không có sự cộng hưởng: Giả thuyết về "Luxury Cluster" (càng nhiều tiện ích giá càng cao) đã bị bác bỏ. Việc sở hữu một "combo" đầy đủ tiện ích không hề đẩy giá trị trung bình của bất động sản lên cao hơn.

```python
avg_luxury_score_by_age = df.groupby('age_category')['Luxury_Score'].mean().reset_index()
print(avg_luxury_score_by_age)
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='age_category', y='Luxury_Score', data=avg_luxury_score_by_age, palette='viridis', hue='age_category', legend=False)
plt.title('Average Luxury Score by Property Age Category')
plt.xlabel('Age Category')
plt.ylabel('Average Luxury Score')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

-> chứng minh được rằng những ngôi nhà mới thường có nhiều tiện ích hơn.


```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='age_category', y='price_per_squareMeter', palette='viridis', alpha=0.7, data=df)
plt.title('Price per Square Meter Distribution by Property Age Category')
plt.xlabel('age_category')
plt.ylabel('Price per Square Meter')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
